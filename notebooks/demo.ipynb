# ChatGPT-style surrogate demo

Asks an open question that requires the web, then shows the **full trace** so you can see every prompt, every tool call, and every tool result the model saw.

Prereqs (run once, in a terminal):
```bash
# tunnel to the GPU box
ssh -fN -L 8000:127.0.0.1:8000 flow-blonde-panther
# venv
source .venv/bin/activate
```

````python
import sys, os
sys.path.insert(0, os.path.abspath('..'))

from surrogate.agent import chat
from surrogate.replay import replay, load
````

````python
answer, msgs, log = chat('Which restaurant is the best in Oxford for steak?')
log.close()
print(answer)
print('\nTrace dir:', log.dir)
````

## Full trace (transcript.md)
Every llm_request, llm_response, tool_call and tool_result, in order.

````python
from IPython.display import Markdown
Markdown(replay(log.dir))
````

## Machine-readable trace (trace.jsonl)
Each event is one JSON line — easy to parse, filter, or replay.

````python
events = load(log.dir)
print(f'{len(events)} events:')
for e in events:
    print(f"  #{e['idx']:>2}  {e['kind']:<16}")
````
